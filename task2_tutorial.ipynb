## Object-oriented programming

Object-oriented programming is a design paradigm for creating programs. It is a popular paradigm for software development because the structure allows for reusable and secure code. It achieves these properties by using templates called classes to create objects that interact with each other. Classes contain variables and functions (methods) which they pass on to specific instances or objects. There are four key principles to create object-oriented programs: abstraction, encapsulation, inheritance and polymorphism. Of these four principles polymorphism is the keystone as it forms the basis of object-oriented programming. This tutorial showcases inheritance and one type of polymorphism using Python.

### Inheritance
Inheritance is a principle of object-orientated programming that reduces the repetition of code. It achieves this by having child classes that are derived from a general or parent class. Child classes inherit all the variables and methods of the parent class while adding variables or methods to create more specific objects. Consider how a bank may construct the various types of bank accounts available to customers. All bank accounts will have common properties, such as a balance, and the ability to deposit and withdraw money from the account. The following Python code defines a parent class for creating a new bank account. When a new bank account (object) is created it will have an initial balance = 0 unless otherwise specified and you will be able to call the `credit()`, `debit()`, and `view()` methods as shown.

```python
# Bank account class (parent class)
class BankAccount:
    def __init__(self, balance=0):
        '''Initilize new account'''
        self.balance = balance

    def credit(self, amount):
        '''Credit account with this amount'''
        self.balance += amount
        print(f"\n Credit: {amount}")

    def debit(self, amount):
       '''Debit account with this amount if there is a sufficient balance'''
       if self.balance>=amount:
            self.balance-=amount
            print("\n You Withdrew:", amount)
       else:
            print("\n Insufficient balance  ")

    def view(self):
         print("\n Balance = ", self.balance)

# Open a new bank account (create object)
account_1 = BankAccount(1000)

# Deposit into account and view balance
account_1.credit(500)
account_1.view()

# Withdraw from account and view balance
account_1.debit(50)
account_1.view()
```

>**Pause and reflect:** Consider what other information or method could be included in the parent `BankAccount()` class.  

Next, the bank would like to create a savings account where customers can earn interest. Since the `BankAccount()` class already has the basic properties required for a bank account we create a `Savings()` class based on the `BankAccount()` class. In addition to the opening balance, the interest rate can also be passed as an argument when creating an object with the `Savings()` class. An additional method is also added to calculate the interest earned. Note how the methods `debit()`, `credit()`, and `view()` can be called by the object `account_2` even though these methods are not defined under the `Savings()` class.


```python
# Savings account class (child class)
class Savings(BankAccount):

    def __init__(self, balance=0, interest_rate=0.05):
        '''Initilize new account with zero balance'''
        super().__init__(balance)
        self.interest_rate = interest_rate

    def compound_monthly(self):
        '''Add interest to balance'''
        self.balance = round(self.balance * (1 + self.interest_rate / 12), 2)


# Open a savings account
account_2 = Savings(1000)
account_2.view()

# Deposit into account and view balance
account_2.credit(500)
account_2.view()

# Withdraw from account and view balance
account_2.debit(500)
account_2.view()

# Balance + interest after one month
account_2.compound_monthly()
account_2.view()
```

### Polymorphism
Polymorphism is the ability of a method to behave differently based on the object it is interacting with. This principle allows for flexibility in a program. One type of polymorphism, overriding, acts by changing the body of the method in a new class but keeping the same method name.  For example, if the bank would like to provide accounts with an overdraft facility the `debit()` method can be modified within the `Overdraft()` child class to allow the balance to become negative. This means when the `debit()` method is called the behaviour of the object will depend on whether it was created using the `BankAccount()` or `Overdraft()` class.

```python
# Overdraft account (child class with polymorphism)
class Overdraft (BankAccount):
    def debit(self, amount):
       '''Debit account with this amount if there is a sufficient balance
       or overdraft facility'''
       if self.balance>=-1000:
            self.balance-=amount
            print("\n You Withdrew:", amount)
       else:
            print("\n Overdraft exceeded")

# Open a overdraft account
account_3 = Overdraft(1000)

# Deposit into the account
account_3.credit(500)
account_3.view()

# Withdraw from account and view balance
account_3.debit(2000)
account_3.view()
```

Overriding does not need to be implemented within an inheritance structure. Consider the `len()` method in Python that determines the length of an object. You can call this method on a string or list, however the body of the method to determine the length of string versus a list is different.
